```python
# ==========================================
# Phase 2: Environment Setup & Data Loading
# ==========================================

# Task 1: Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Load Titanic dataset
# Upload train.csv to Colab before running
df = pd.read_csv("/content/Titanic-Dataset.csv")

print("Dataset Loaded Successfully!")
```

```python
# ==========================================
# Phase 3: Data Inspection
# ==========================================

# Task 3: Display first 10 rows
print("First 10 Rows")
df.head(10)


# Task 4: Display last 5 rows
print("Last 5 Rows")
df.tail(5)


# Task 5: Check dataset dimensions
print("Shape of Dataset:")
print(df.shape)


# Task 6: Display dataset information
print("Dataset Information")
df.info()


# Task 7: Summary statistics
print("Summary Statistics")
print(df.describe())

print("\nAge Statistics")
print("Mean:", df["Age"].mean())
print("Minimum:", df["Age"].min())
print("Maximum:", df["Age"].max())

print("\nFare Statistics")
print("Mean:", df["Fare"].mean())
print("Minimum:", df["Fare"].min())
print("Maximum:", df["Fare"].max())


# Task 8: Missing values
print("Missing Values")
print(df.isnull().sum())
```

```python
# ==========================================
# Phase 4: Data Manipulation & Cleaning
# ==========================================

# Task 9: Fill missing Age values with median
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("Missing Age values filled with median.")

# Task 10: Filter surviving passengers
survivors = df[df["Survived"] == 1]

print("Number of Survivors:", len(survivors))

survivors.head()

# Task 11: Convert Fare column to NumPy array
fare_array = df["Fare"].to_numpy()

print(fare_array)

# Task 12: Maximum and Average Fare
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print("Maximum Fare:", max_fare)
print("Average Fare:", avg_fare)
```

```python
# ==========================================
# Phase 5: Data Visualization
# ==========================================


#Task 13: Histogram of Age
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# Task 14: Bar Chart of Survival Count
plt.figure(figsize=(6,4))

survival_counts = df["Survived"].value_counts()

plt.bar(["Did Not Survive", "Survived"], survival_counts)

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")

plt.show()

# Task 15: Scatter Plot of Age vs Fare
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Fare"])

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```
